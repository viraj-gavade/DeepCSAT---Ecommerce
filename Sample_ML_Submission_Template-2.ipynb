## 3. ***Data Wrangling***

````python
df.shape
````

### Data Wrangling Code

````python
import pandas as pd
import numpy as np

# ==============================
# 1. Load Dataset
# ==============================
df = pd.read_csv("eCommerce_Customer_support_data.csv")

print("Initial Shape:", df.shape)

# ==============================
# 2. Drop Irrelevant Columns
# ==============================
df.drop(columns=['Unique id', 'Order_id'], inplace=True)

# ==============================
# 3. Convert Date Columns
# ==============================
date_columns = [
    'order_date_time',
    'Issue_reported at',
    'issue_responded',
    'Survey_response_Date'
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors='coerce')

# ==============================
# 4. Create New Time Features
# ==============================
df['order_month'] = df['order_date_time'].dt.month
df['order_day'] = df['order_date_time'].dt.day
df['order_hour'] = df['order_date_time'].dt.hour

# Response Time Feature
df['response_time_minutes'] = (
    df['issue_responded'] - df['Issue_reported at']
).dt.total_seconds() / 60

# ==============================
# 5. Handle Missing Values
# ==============================

# Numerical columns
df['connected_handling_time'].fillna(
    df['connected_handling_time'].median(),
    inplace=True
)

df['Item_price'].fillna(
    df['Item_price'].median(),
    inplace=True
)

df['response_time_minutes'].fillna(
    df['response_time_minutes'].median(),
    inplace=True
)

df['order_date_time'].fillna(
    df['order_date_time'].mode()[0],
    inplace=True
)

df['Issue_reported at'].fillna(
    df['Issue_reported at'].mode()[0],
    inplace=True
)

df['issue_responded'].fillna(
    df['issue_responded'].mode()[0],
    inplace=True
)


df['order_month'].fillna(
    df['order_month'].mode()[0],
    inplace=True
)

df['order_day'].fillna(
    df['order_day'].mode()[0],
    inplace=True
)
df['order_hour'].fillna(
    df['order_hour'].mode()[0],
    inplace=True
)

# Categorical columns
categorical_cols = [
    'Customer_City',
    'Product_category',
    'Customer Remarks',
]

for col in categorical_cols:
    df[col].fillna("Unknown", inplace=True)

# ==============================
# 6. Clean Text Column
# ==============================
df['Customer Remarks'] = (
    df['Customer Remarks']
    .str.lower()
    .str.strip()
)

# ==============================
# 7. Remove Extreme Outliers
# ==============================
df = df[df['Item_price'] < df['Item_price'].quantile(0.99)]
df = df[df['connected_handling_time'] < df['connected_handling_time'].quantile(0.99)]

# ==============================
# 8. Final Check
# ==============================
print("Final Shape:", df.shape)
print("\nRemaining Missing Values:\n", df.isnull().sum())
````
