### Data Wrangling Code

````python
import pandas as pd
import numpy as np

# ==============================
# 1️⃣ Load Dataset
# ==============================
# df = pd.read_csv("eCommerce_Customer_support_data.csv")

print("Initial Shape:", df.shape)


# ==============================
# 2️⃣ Drop Irrelevant Columns
# ==============================
cols_to_drop = ['Unique id', 'Order_id']
df = df.drop(columns=[col for col in cols_to_drop if col in df.columns])


# ==============================
# 3️⃣ Convert Date Columns
# ==============================
date_columns = [
    'order_date_time',
    'Issue_reported at',
    'issue_responded',
    'Survey_response_Date'
]

for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors='coerce')


# ==============================
# 4️⃣ Create Time-Based Features
# ==============================
df['order_month'] = df['order_date_time'].dt.month
df['order_day'] = df['order_date_time'].dt.day
df['order_hour'] = df['order_date_time'].dt.hour

# Response Time (in minutes)
df['response_time_minutes'] = (
    df['issue_responded'] - df['Issue_reported at']
).dt.total_seconds() / 60


# ==============================
# 5️⃣ Handle Missing Values
# ==============================

# Numerical Columns
numerical_cols = [
    'connected_handling_time',
    'Item_price',
    'response_time_minutes',
    'order_month',
    'order_day',
    'order_hour'
]

for col in numerical_cols:
    if col in df.columns:
        df[col].fillna(df[col].median(), inplace=True)

# Date Columns (fill with mode)
for col in ['order_date_time', 'Issue_reported at', 'issue_responded']:
    if col in df.columns:
        df[col].fillna(df[col].mode()[0], inplace=True)

# Categorical Columns
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    print(f"Filling missing values in '{col}' with 'Unknown'")
    df[col].fillna("Unknown", inplace=True)
    
                


# ==============================
# 6️⃣ Clean Text Column
# ==============================
if 'Customer Remarks' in df.columns:
    df['Customer Remarks'] = (
        df['Customer Remarks']
        .str.lower()
        .str.strip()
    )


# ==============================
# 7️⃣ Handle Outliers (IQR Capping)
# ==============================

numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Exclude target variable
numeric_cols = [col for col in numeric_cols if col != 'CSAT Score']

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    df[col] = df[col].clip(lower, upper)


# ==============================
# 8️⃣ Final Verification
# ==============================

print("\nFinal Shape:", df.shape)

print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nData Types:")
print(df.dtypes)
````

# 📌 Outlier Treatment Techniques Used

## 1️⃣ Interquartile Range (IQR) Method for Detection

We used the **Interquartile Range (IQR)** method to detect outliers in numerical columns.

The IQR method identifies outliers using the following formulas:

Lower Bound = Q1 − 1.5 × IQR  
Upper Bound = Q3 + 1.5 × IQR  

Any value falling outside this range is considered an **outlier**.

### ✅ Why IQR?

- It is **robust to skewed data**
- It does **not assume normal distribution**
- It is suitable for **small datasets**
- It works well for operational metrics like **response time** and **item price**

---

## 2️⃣ IQR-Based Capping (Winsorization)

Instead of removing rows containing outliers, we applied **IQR-based capping** using the `clip()` function.

Extreme values were replaced with the calculated **upper and lower bounds**.

### ✅ Why Capping Instead of Removing?

- The dataset size is small (**120 rows**)
- Removing outliers would significantly reduce training data
- Many extreme values represent **real business scenarios** (e.g., high response time, high item price)
- Artificial Neural Networks handle scaled extreme values better than linear models
- Capping preserves dataset size while reducing the influence of extreme values

---

## 3️⃣ Exclusion of Target Variable from Outlier Treatment

The target variable **CSAT Score** was not modified, even though the IQR method flagged some values as outliers.

### ✅ Why?

- CSAT scores (**1–5**) are valid categorical labels
- Removing or altering them would distort the prediction task
- Target variable integrity must be preserved in supervised learning

---

# 📌 Final Conclusion on Outlier Handling

Outliers were detected using the **IQR method** and treated using **capping instead of deletion**.

This approach ensured:

- Data integrity  
- Preservation of sample size  
- Minimal distortion during model training

````python
# Select your features wisely to avoid overfitting
import pandas as pd
import numpy as np
from sklearn.feature_selection import VarianceThreshold, SelectKBest, f_classif

# ===============================
# 1️⃣ Remove Low Variance Features
# ===============================

variance_filter = VarianceThreshold(threshold=0.01)
X_var = variance_filter.fit_transform(X)

# Keep selected columns
selected_columns_var = X.columns[variance_filter.get_support()]
X = pd.DataFrame(X_var, columns=selected_columns_var)

print("After Variance Threshold:", X.shape)
print(X.isnull().sum())
print("Total NaNs in X:", X.isnull().sum().sum())

# ===============================
# 2️⃣ Correlation With Target
# ===============================

# Combine temporarily for correlation check
temp_df = X.copy()
temp_df['CSAT Score'] = y

corr_with_target = temp_df.corr()['CSAT Score'].abs().sort_values(ascending=False)

print("\nCorrelation with Target:")
print(corr_with_target)

# Keep features with meaningful correlation (> 0.02)
selected_features_corr = corr_with_target[corr_with_target > 0.02].index
selected_features_corr = selected_features_corr.drop('CSAT Score')

X = X[selected_features_corr]

print("\nAfter Correlation Filtering:", X.shape)


# ===============================
# 3️⃣ Select Top K Features (Optional but Recommended)
# ===============================

# Select top 20 best features
selector = SelectKBest(score_func=f_classif, k=min(20, X.shape[1]))
X_new = selector.fit_transform(X, y)

selected_columns_kbest = X.columns[selector.get_support()]
X = pd.DataFrame(X_new, columns=selected_columns_kbest)

print("\nFinal Shape After SelectKBest:", X.shape)
````

## ***7. ML Model Implementation***

### ML Model - 1

````python
import tensorflow as tf
tf.keras.backend.clear_session()
````

````python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

# Adjust labels to zero-based
y_train_zero = y_train - 1

classes = np.unique(y_train_zero)

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=classes,
    y=y_train_zero
)

class_weight_dict = dict(zip(classes, class_weights))

print("Class Weights:", class_weight_dict)
````

````python
from tensorflow import keras
from tensorflow.keras import layers

model = keras.Sequential([
    
    layers.Dense(128, activation='relu', input_shape=(X_train.shape[1],)),
    layers.Dropout(0.3),
    
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.3),
    
    layers.Dense(len(classes), activation='softmax')
])
````

````python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
````

````python
history = model.fit(
    X_train,
    y_train_zero,
    epochs=25,
    batch_size=64,
    validation_split=0.2,
    class_weight=class_weight_dict,
    verbose=1
)
````
